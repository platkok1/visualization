```python
# import necessary libraries 
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd 
```

```python
np.random.seed(613) # Sets the seed for NumPy's random number generator to ensure reproducibility
x = np.arange(50) # x is is the x-axis coordinates 
y = np.random.randint(0, 100, 50) # Generates an array y of 50 random integers between 0 100 
fig, ax = plt.subplots(figsize = (5,3)) # figure size of 5 inches wide and 3 inches tall
#ax.scatter(x,y)
#ax.bar (x,y)

# The activity classs was asked to modify code to produce a histogram 
#The class code did not include expliciot binning size. It wa automatically generated 
# ax.hist(y, bins=10)  my original code djusted 'bins' for different bin sizes 
ax.hist(y)
```

```python
# Define a font style dictionary
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

# Create a new figure and axes object with size 5 inches by 3 inches 
fig, ax = plt.subplots(figsize=(5, 3))
# Plot y vs. x 
ax.plot(x,y)

# Set custom styping for title, x- and y- axis 
ax.set_title('Total growth over time', fontdict = font1,loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
# Create a new figure 
fig, ax = plt.subplots(figsize=(5, 3))

# Plot a scatter plot with custom paramaters for markers 

ax.scatter( x,
            y,
            marker='*', 
            color = "indigo") 
fig.show()
```

```python
# Create a new figure 
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot( x,
            y,
            marker='*', #Marker type
            color = "indigo", # Marker colour 
            linestyle = '--',# Line style 
            linewidth = 2 #Specify line width 
            ) 
fig.show()
```

```python
#Create a new plot based on the code above and add more customization to it 
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='*',
        markersize = 12, 
        color = '#7425b9', 
        linestyle = '--' ,
        linewidth = 2,
        markeredgecolor = '#fa9359', 
        markerfacecolor = '#fa9359') 
ax.grid (axis = 'y', color = "red", linewidth = 2, linestyle = '-.')
fig.show()
```

```python
# Activity 
# Library imported previsously 
# Link to original visualization https://python-graph-gallery.com/10-barplot-with-number-of-observation/

# Create bars
barWidth = 1.5
bars1 = [3, 3, 1]
bars2 = [4, 2, 3]
bars3 = [4, 6, 7, 10, 4, 4]
bars4 = bars1 + bars2 + bars3
 
# The X position of bars
r1 = [1,5,9] #Customized bar width 
r2 = [2,6,10]
r3 = [3,4,7,8,11,12]
r4 = r1 + r2 + r3
 
# Create barplot
plt.bar(r1, bars1, width = barWidth, color = (0.3,0.1,0.4,0.1), label='Alone') # Customized colour 
plt.bar(r2, bars2, width = barWidth, color = (0.3,0.5,0.4,0.6), label='With Himself')
plt.bar(r3, bars3, width = barWidth, color = (0.3,0.9,0.4,0.6), label='With other genotype')

# Note: the barplot could be created easily. See the barplot section for other examples.
 
# Create legend
plt.legend()
 
# Text below each barplot with a rotation at 90°
plt.xticks([r + barWidth for r in range(len(r4))], ['DD', 'with himself', 'with DC', 'with Silur', 'DC', 'with himself', 'with DD', 'with Silur', 'Silur', 'with himself', 'with DD', 'with DC'], rotation=90)
 
# Create labels
label = ['n = 6', 'n = 25', 'n = 13', 'n = 36', 'n = 30', 'n = 11', 'n = 16', 'n = 37', 'n = 14', 'n = 4', 'n = 31', 'n = 34']
 
# Text on the top of each bar
for i in range(len(r4)):
    plt.text(x = r4[i]-0.5 , y = bars4[i]+0.1, s = label[i], size = 6)

# Adjust the margins
plt.subplots_adjust(bottom= 0.2, top = 0.98)
 
# Show graphic
plt.show()
```
