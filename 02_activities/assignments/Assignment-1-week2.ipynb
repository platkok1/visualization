Accessibility activity 

Level 1 – Basic Descriptive Observations
Focus: Just the facts. What do we see in the chart?

- Identifying the chart type (bar chart, scatterplot, etc.)

- Describing the axes (e.g., time on the x-axis, hours of activity on the y-axis)

- Mentioning units and ranges (e.g., hours per day, categories like "sleep", "work", "exercise")

- Identifying colours and what they represent (e.g., red = sleep, green = work)

- Mentioning number of individuals/data points (e.g., "each row is one person")

- No interpretation, just structured description

Example: “This bar shows 6 hours of sleep, represented in blue, occurring between midnight and 6 AM.”
Not very useful 

Level 2 – Pattern Recognition and Categorization
Focus: Grouping, summarizing, and lightly interpreting patterns within the data.

- Observing common trends (e.g., “many people slept less when they were working on creative projects”)

- Categorizing routines or daily patterns (e.g., “people had balanced schedules with sleep, work, and meals”)

- Noting differences between individuals or groups

Example: “Some individuals showed consistent daytime work patterns, while others were more active at night.”
Valuable for making conclusions about the presented data 

Level 3 – Interpretation and Inference
Focus: Making meaning from the data itself. Drawing conclusions based only on what’s in the chart.

- Inferring lifestyles or habits based on observed patterns

- Making comparisons between people/groups (e.g., “Most had enough sleep and food time; only a few prioritized exercise.”)

- Pointing out anomalies or exceptions (e.g., “One individual consistently had irregular sleep patterns.”)

Example: “Creative individuals appeared to cluster their work at night, suggesting a preference for nighttime productivity.”


Level 4 – External Context and Critical Reflection
Focus: Bringing in outside knowledge or social/historical context to enrich interpretation.

- Integrating knowledge of historical or social factors (e.g., “Many of these figures lived in the 1800s when structured day jobs were less common.”)

- Discussing cultural or societal expectations (e.g., “Women’s schedules may reflect constraints like childcare.”)

- Including external assumptions, scholarly insight, or opinion (e.g., “We often assume creative people work at night; this data partially supports that.”)

- Critical commentary (e.g., “These routines reflect class privilege, as not everyone could afford a flexible day.”)

Example: “The absence of day jobs among most individuals suggests they may have belonged to a privileged social class with financial freedom.”

Impacts of Bias:

```python
#Import plotly 
import plotly.graph_objects as go 

#Generate sample data 
x1 = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1 = np.array([110,180, 240, 99, 220])

#Make figure 
graph = go.Figure()

#Choose plot type to be bar graph 
graph.add_trace(go.Bar(x=x1, y=y1))

# Include a tytle 
graph.update_layout(
    title = 'Pirate Stores',
    xaxis_title = 'Pirates',
    yaxis_title = 'Score'
)

# Add scatter trace to bar graph 
graph.add_trace(go.Scatter(
    x=x1,
    y=y1,
    mode='markers',
    marker=dict(
        size=15,
        color='hotpink',
        opacity=1,
        line=dict(width=5, color='purple')
    )
))

# Update layout correctly
graph.update_layout(
    title='Pirate Scores',
    xaxis=dict(title='Pirates'),
    yaxis=dict(title='Scores'),
    width=500,
    height=500
)
#Show graph 
graph.show()
```
